### Feature Engineering
##### Feature engineering means creating or transforming features from existing data to help a machine learning model identify useful patterns.

```python
# Check the Distribution of data
print("Amount's Skew Value:", data['Amount'].skew())

#Create the new feature using the log transformation
data['Amount_log'] = np.log1p(data['Amount'])
# np.log1p(x) calculates ln(1 + x) to compress large numbers and reduce skewness. For example, an amount of 100,000 becomes approximately 11.51, while 0 becomes 0.

print("Amount_log skewness:", data['Amount_log'].skew())

#Check the new Data
print(data[['Amount', 'Amount_log']].head())
```

# UnSupervised - Anamoly Deduction

Select appropriate transaction features

```python
# Create a separate feature dataset for anomaly detection
X_anomaly = data.drop(columns=['Class', 'Amount_log'])

# Check the selected features
print(X_anomaly.columns)
```

```python
# Check Missing Values 
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy="mean")

X_imputed = pd.DataFrame(
    imputer.fit_transform(X_anomaly),
    columns=X_anomaly.columns,
    index=X_anomaly.index
)

print("\nMissing Values After Imputation:")
print(X_imputed.isnull().sum())
```

```python
# Use Standard Scaler
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train_Anamoly = scaler.fit_transform(X_anomaly)
print(x_train_Anamoly)
```

Isolation Model Creation

```python
# Model Creation  
from sklearn.ensemble import IsolationForest

Iso_model = IsolationForest(
        n_estimators= 100,
        contamination='auto' ,
        random_state=42
)

Iso_model.fit(X_imputed)

Iso_model_predict = Iso_model.predict(X_imputed)
Iso_model_predict
```

```python
from sklearn.metrics import (
    precision_score, recall_score, f1_score,
    classification_report, confusion_matrix
)

# Actual fraud labels
y_true = data['Class'] #From the data it has the prebuilt Fraud values

# Convert Isolation Forest predictions: -1 = anomaly, 1 = normal
y_pred = (Iso_model_predict == -1).astype(int)

#zero_division=0 means if precision, recall, or F1-score involves division by zero, return 0 instead of raising a warning or error. 
Iso_pre = precision_score(y_true, y_pred, zero_division=0)
Iso_rec = recall_score(y_true, y_pred, zero_division=0)
Iso_f1 = f1_score(y_true, y_pred, zero_division=0)
Iso_class_rep = classification_report(y_true, y_pred, zero_division=0)

print('Precision_score:', Iso_pre)
print('Recall_score:', Iso_rec)
print('F1_score:', Iso_f1)
print('Classification_report:\n', Iso_class_rep)
print('Confusion Matrix:\n', confusion_matrix(y_true, y_pred))

print('y_true','\n', y_true)
print(y_pred)
```

```python
Iso_anomaly_score = Iso_model.decision_function(X_imputed)
print(Iso_anomaly_score)
```

Local Outlier Factor

```python
from sklearn.neighbors import LocalOutlierFactor

#This process is for reduce the running time 
LOF_sample = X_imputed.sample(n=100, random_state=42)

LOF_scaler = StandardScaler()
LOF_Scaled = LOF_scaler.fit_transform(LOF_sample)

LOF = LocalOutlierFactor(
    n_neighbors= 5,
    contamination= 'auto', 
    novelty=True,
    n_jobs=-1
)

LOF.fit(LOF_Scaled)

lof_predict = LOF.predict(LOF_Scaled)

# Predict Anomaly Score
lof_anomaly_score = LOF.decision_function(LOF_Scaled)
print(lof_anomaly_score)
```

```python
from sklearn.metrics import (
    precision_score, recall_score, f1_score,
    classification_report, confusion_matrix
)

# Actual fraud labels
y_true = data.loc[LOF_sample.index, 'Class']#From the data it has the prebuilt Fraud values

# Convert Isolation Forest predictions: -1 = anomaly, 1 = normal
y_pred = (lof_predict == -1).astype(int)

#zero_division=0 means if precision, recall, or F1-score involves division by zero, return 0 instead of raising a warning or error. 
LOF_pre = precision_score(y_true, y_pred, zero_division=0)
LOF_rec = recall_score(y_true, y_pred, zero_division=0)
LOF_f1 = f1_score(y_true, y_pred, zero_division=0)
LOF_class_rep = classification_report(y_true, y_pred, zero_division=0)

print('Precision_score:', LOF_pre)
print('Recall_score:', LOF_rec)
print('F1_score:', LOF_f1)
print('Classification_report:\n', LOF_class_rep)
print('Confusion Matrix:\n', confusion_matrix(y_true, y_pred))

print('y_true','\n', y_true)
print(y_pred)
```

```python
from sklearn.svm import OneClassSVM
from sklearn.preprocessing import StandardScaler

X_sample = LOF_sample.copy()

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_sample)

svm_model = OneClassSVM(
    kernel='rbf',
    nu=0.01,
    gamma='scale'
)

svm_model.fit(X_scaled)

svm_predict = svm_model.predict(X_scaled)
svm_anomaly_score = svm_model.decision_function(X_scaled)

print(svm_anomaly_score)
```

## Final Evaluation using Hybrid Model

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Select test transactions for anomaly detection
X_test_anomaly = X_imputed.loc[X_test.index] #.loc == is used in Pandas to select rows and columns using their labels

# Get Isolation Forest anomaly scores for test data
Iso_test_score = Iso_model.decision_function(X_test_anomaly)

# Convert anomaly scores into 0-1 scale
Iso_anomaly_normalized = iso_score_scaler.transform(
    (-Iso_test_score).reshape(-1, 1)
).flatten()

# Create Hybrid DataFrame
hybrid_data = pd.DataFrame(index=X_test.index)

hybrid_data['classification_score'] = xgb_prob
hybrid_data['Iso_score'] = Iso_anomaly_normalized

# Combine both scores
hybrid_data['hybrid_score'] = (
    hybrid_data['classification_score'] * 0.5 +
    hybrid_data['Iso_score'] * 0.5
)

# Generate hybrid predictions
hybrid_pred = (hybrid_data['hybrid_score'] >= 0.5).astype(int)

print(hybrid_data.head())
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

# Classification predictions
classification_pred = xgb_pred

# Anomaly predictions
anomaly_pred = (Iso_model.predict(X_test_anomaly) == -1).astype(int)

# Actual test labels
y_true = y_test

# Comparison
comparison = pd.DataFrame({
    'Algorithm': [
        'Classification',
        'Isolation Forest',
        'Hybrid Model'
    ],
    'Precision': [
        precision_score(y_true, classification_pred, zero_division=0),
        precision_score(y_true, anomaly_pred, zero_division=0),
        precision_score(y_true, hybrid_pred, zero_division=0)
    ],
    'Recall': [
        recall_score(y_true, classification_pred, zero_division=0),
        recall_score(y_true, anomaly_pred, zero_division=0),
        recall_score(y_true, hybrid_pred, zero_division=0)
    ],
    'F1-Score': [
        f1_score(y_true, classification_pred, zero_division=0),
        f1_score(y_true, anomaly_pred, zero_division=0),
        f1_score(y_true, hybrid_pred, zero_division=0)
    ]
})

print(comparison)
```

```python
def risk_category(score):
    if score >= 0.75:
        return 'Critical'
    elif score >= 0.50:
        return 'High'
    elif score >= 0.25:
        return 'Medium'
    else:
        return 'Low'

hybrid_data['Risk_Category'] = hybrid_data['hybrid_score'].apply(risk_category)

print(hybrid_data['Risk_Category'].value_counts())
```

```python
suspicious_transactions = hybrid_data[
    hybrid_data['Risk_Category'].isin(['High', 'Critical'])
]

print("Total Suspicious Transactions:", len(suspicious_transactions))
print(suspicious_transactions.head())
```

```python
from sklearn.metrics import confusion_matrix

hybrid_pred = (hybrid_data['hybrid_score'] >= 0.50).astype(int)

FP = ((y_test == 0) & (hybrid_pred == 1)).sum()
FN = ((y_test == 1) & (hybrid_pred == 0)).sum()

print("False Positives:", FP)
print("False Negatives:", FN)
```

```python

# ==================================================
# STEP 1: IMPORT LIBRARIES
# ==================================================

import pandas as pd
import numpy as np
import re
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import wordpunct_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.util import ngrams

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# Download required NLTK resources
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')


# ==================================================
# STEP 2: CREATE DATASET
# ==================================================

data = {
    "review": [
        "This product is absolutely amazing",
        "I really love this product",
        "Excellent quality and great performance",
        "The product is very useful",
        "Amazing experience with this product",
        "I am happy with my purchase",
        "This is a wonderful product",
        "The quality is excellent",
        "Very good product and fast delivery",
        "I highly recommend this product",
        "Fantastic product and great service",
        "The product works perfectly",
        "I am satisfied with the quality",
        "Really nice and useful product",
        "Best purchase I have ever made",
        "Good quality and excellent service",
        "The experience was wonderful",
        "I absolutely enjoy using this product",
        "This product is worth buying",
        "Great product with amazing features",

        "This product is very bad",
        "I hate this product",
        "Terrible quality and poor performance",
        "The product is completely useless",
        "Very disappointing experience",
        "I am unhappy with my purchase",
        "This is a horrible product",
        "The quality is very poor",
        "Bad product and slow delivery",
        "I do not recommend this product",
        "Awful product and terrible service",
        "The product does not work",
        "I am disappointed with the quality",
        "Really bad and useless product",
        "Worst purchase I have ever made",
        "Poor quality and bad service",
        "The experience was terrible",
        "I regret buying this product",
        "This product is not worth buying",
        "Very poor product with bad features"
    ],

    "sentiment": [
        "Positive", "Positive", "Positive", "Positive",
        "Positive", "Positive", "Positive", "Positive",
        "Positive", "Positive", "Positive", "Positive",
        "Positive", "Positive", "Positive", "Positive",
        "Positive", "Positive", "Positive", "Positive",

        "Negative", "Negative", "Negative", "Negative",
        "Negative", "Negative", "Negative", "Negative",
        "Negative", "Negative", "Negative", "Negative",
        "Negative", "Negative", "Negative", "Negative",
        "Negative", "Negative", "Negative", "Negative"
    ]
}

df = pd.DataFrame(data)

print("DATASET:")
print(df.head())

print("\nDataset Shape:", df.shape)


# ==================================================
# STEP 3: INITIALIZE NLP TOOLS
# ==================================================

stop_words = set(stopwords.words('english'))

stemmer = PorterStemmer()

lemmatizer = WordNetLemmatizer()


# ==================================================
# STEP 4: TEXT PREPROCESSING FUNCTION
# ==================================================

def preprocess_text(text):

    # 1. Text Cleaning
    text = re.sub(r'http\S+|www\S+', '', text)
    text = re.sub(r'<.*?>', '', text)

    # 2. Lowercasing
    text = text.lower()

    # 3. Punctuation Removal
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)

    # 4. Tokenization
    tokens = wordpunct_tokenize(text)

    # 5. Stopword Removal
    tokens = [
        word for word in tokens
        if word not in stop_words
    ]

    # 6. Stemming
    stemmed_words = [
        stemmer.stem(word)
        for word in tokens
    ]

    # 7. Lemmatization
    lemmatized_words = [
        lemmatizer.lemmatize(word)
        for word in tokens
    ]

    return tokens, stemmed_words, lemmatized_words


# ==================================================
# STEP 5: APPLY PREPROCESSING
# ==================================================

df[['tokens', 'stemmed', 'lemmatized']] = df['review'].apply(
    lambda text: pd.Series(preprocess_text(text))
)

print("\nPREPROCESSING OUTPUT:")
print(df[['review', 'tokens', 'stemmed', 'lemmatized']].head())


# ==================================================
# STEP 6: N-GRAMS
# ==================================================

sample_tokens = df['lemmatized'].iloc[0]

unigrams = list(ngrams(sample_tokens, 1))
bigrams = list(ngrams(sample_tokens, 2))
trigrams = list(ngrams(sample_tokens, 3))

print("\nUNIGRAMS:")
print(unigrams)

print("\nBIGRAMS:")
print(bigrams)

print("\nTRIGRAMS:")
print(trigrams)


# ==================================================
# STEP 7: PREPARE FINAL TEXT
# ==================================================

df['processed_text'] = df['lemmatized'].apply(
    lambda words: ' '.join(words)
)

X = df['processed_text']
y = df['sentiment']


# ==================================================
# STEP 8: TRAIN TEST SPLIT
# ==================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("\nTraining Samples:", len(X_train))
print("Testing Samples:", len(X_test))


# ==================================================
# STEP 9: BAG OF WORDS (BoW)
# ==================================================

bow_vectorizer = CountVectorizer(
    ngram_range=(1, 1)
)

X_train_bow = bow_vectorizer.fit_transform(X_train)

X_test_bow = bow_vectorizer.transform(X_test)

print("\nBoW Vocabulary:")
print(bow_vectorizer.get_feature_names_out())

print("\nBoW Training Shape:", X_train_bow.shape)
print("BoW Testing Shape:", X_test_bow.shape)


# ==================================================
# STEP 10: TF-IDF VECTORIZATION
# ==================================================

tfidf_vectorizer = TfidfVectorizer(
    ngram_range=(1, 1)
)

X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)

X_test_tfidf = tfidf_vectorizer.transform(X_test)

print("\nTF-IDF Vocabulary:")
print(tfidf_vectorizer.get_feature_names_out())

print("\nTF-IDF Training Shape:", X_train_tfidf.shape)
print("TF-IDF Testing Shape:", X_test_tfidf.shape)


# ==================================================
# STEP 11: TRAIN MODEL USING BoW
# ==================================================

bow_model = LogisticRegression(max_iter=1000)

bow_model.fit(X_train_bow, y_train)

bow_predictions = bow_model.predict(X_test_bow)


# ==================================================
# STEP 12: EVALUATE BoW MODEL
# ==================================================

print("\n========== BoW EVALUATION ==========")

print("Accuracy:",
      accuracy_score(y_test, bow_predictions))

print("\nClassification Report:")
print(classification_report(y_test, bow_predictions))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, bow_predictions))


# ==================================================
# STEP 13: TRAIN MODEL USING TF-IDF
# ==================================================

tfidf_model = LogisticRegression(max_iter=1000)

tfidf_model.fit(X_train_tfidf, y_train)

tfidf_predictions = tfidf_model.predict(X_test_tfidf)


# ==================================================
# STEP 14: EVALUATE TF-IDF MODEL
# ==================================================

print("\n========== TF-IDF EVALUATION ==========")

print("Accuracy:",
      accuracy_score(y_test, tfidf_predictions))

print("\nClassification Report:")
print(classification_report(y_test, tfidf_predictions))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, tfidf_predictions))


# ==================================================
# STEP 15: COMPARE BOTH MODELS
# ==================================================

bow_accuracy = accuracy_score(y_test, bow_predictions)

tfidf_accuracy = accuracy_score(y_test, tfidf_predictions)

comparison = pd.DataFrame({
    "Vectorization": ["BoW", "TF-IDF"],
    "Accuracy": [bow_accuracy, tfidf_accuracy]
})

print("\nMODEL COMPARISON:")
print(comparison)


# ==================================================
# STEP 16: TEST WITH NEW REVIEWS
# ==================================================

new_reviews = [
    "This product is amazing and excellent",
    "Very bad product and poor quality"
]

new_processed = [
    ' '.join(preprocess_text(review)[2])
    for review in new_reviews
]

# Transform using the already fitted vectorizers
new_bow = bow_vectorizer.transform(new_processed)

new_tfidf = tfidf_vectorizer.transform(new_processed)

print("\nNEW REVIEW PREDICTIONS:")

print("BoW Predictions:")
print(bow_model.predict(new_bow))

print("\nTF-IDF Predictions:")
print(tfidf_model.predict(new_tfidf))
```
